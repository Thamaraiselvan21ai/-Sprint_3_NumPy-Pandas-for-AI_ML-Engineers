```python
import pandas as pd
```

```python
#Create a Sample Dataset
students = pd.DataFrame({
    "ID": [101, 102, 103, 104, 105, 106],
    "Name": ["Karan", "Rahul", "Priya", "John", "Meena", "David"],
    "Age": [22, 23, 21, 24, 22, 23],
    "Department": ["AI", "ML", "DS", "AI", "ML", "DS"],
    "Marks": [95, 88, 91, 84, 89, 76]
})

print(students)
```

### 1. loc
loc selects data using row labels (index) and column names.

Syntax:

df.loc[row_label, column_name]

Example:

A school administrator wants to view only the student names and marks from the student database.

### AI/ML Usage
Used to select required features before model training.

Select a Row

```python
students.loc[2]
```

```python
students.loc[:, ["Name", "Marks"]]
```

### 2. iloc
iloc selects data using integer positions instead of labels.

Syntax:

df.iloc[row_position, column_position]

#### Example:

An HR manager wants to view the first three employees and selected columns from an employee dataset.

AI/ML Usage
Useful when column positions are known instead of column names.

```python
students.iloc[0]
```

```python
students.iloc[0:3, 1:4]
```

### Difference Between loc and iloc
loc :

* Uses labels
* Includes end label
* Uses column names
#### iloc:

* Uses integer positions
* Excludes end position like Python slicing
* Uses column numbers

### 3. Boolean Indexing
Boolean Indexing filters rows based on a condition that returns True or False.

Example:

A company wants to find employees earning more than ₹50,000.

#### AI/ML Usage
Used to remove unwanted records before training.

```python
students["Marks"] > 90
```

```python
students[students["Marks"] > 90]
```

### 4. Filtering Rows
Filtering returns only the rows that satisfy a given condition.

Example:

A bank filters customers who have:

* Account balance above ₹1,00,000
* Premium membership
#### AI/ML Usage
Filtering removes unnecessary data before feature engineering.

Students with Marks greater than or equal to 90.

```python
students[students["Marks"] >= 90]
```

```python
#Students from AI Department.
students[students["Department"] == "AI"]
```

```python
#Multiple Conditions

students[
    (students["Department"] == "ML") &
    (students["Marks"] > 85)
]
```

### 5. query()
query() filters rows using SQL-like expressions.

It makes filtering easier to read.

Syntax:

df.query("condition")

#### Example:

A manager searches for employees with more than five years of experience and a salary above ₹60,000.

#### AI/ML Usage
Makes complex filtering conditions easier to write and understand.

```python
students.query("Marks > 90")
```

```python
students.query("Department == 'AI'")
```

```python
students.query("Age >= 22 and Marks >= 85")
```

### 6. isin()
isin() checks whether each value exists in a given list.

Syntax:

df["column"].isin(list)

#### Example:

A company wants to view employees working only in the IT and Finance departments.

#### AI/ML Usage
Useful when filtering multiple categories.

```python
students[
    students["Department"].isin(["AI", "DS"])
]
```

### 7. between()
between() checks whether values fall within a specified range.

Syntax:

df["column"].between(start, end)

#### Example:

A teacher wants to identify students whose marks are between 85 and 92.

#### AI/ML Usage
Used to select records within acceptable ranges, such as age, salary, or income.

```python
students[
    students["Marks"].between(85, 92)
]
```

### 8. Sorting Data
Sorting arranges data in ascending or descending order.

#### Example:

An online shopping website sorts products by price, rating, or popularity.

#### AI/ML Usage
Sorting helps identify top-performing records, detect anomalies, and rank prediction results.

```python
#Sort by Marks
students.sort_values("Marks")
```

```python
students.sort_values(
    "Marks",
    ascending=False
)
```

```python
#Sort by Multiple Columns
students.sort_values(
    ["Department", "Marks"],
    ascending=[True, False]
)
```
