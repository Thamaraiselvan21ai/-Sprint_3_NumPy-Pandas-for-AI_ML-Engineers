### Introduction to Reading and Writing Data
* Reading data means loading data from a file or database into Python.
* Writing data means saving processed data from Python into a file or database.

```python
import pandas as pd
import numpy as np
students = pd.DataFrame({
    "ID": [101, 102, 103, 104, 105],
    "Name": ["Thamarai", "Rahul", "Priya", "John", "Meena"],
    "Age": [22, 23, 21, 24, 22],
    "Department": ["AI", "ML", "DS", "AI", "ML"],
    "Marks": [95, 88, 91, 84, 89]
})

print(students)
```

### 1. CSV Files
CSV (Comma Separated Values) stores tabular data where each value is separated by commas.
It is the most commonly used format in Data Science and Machine Learning.

Example:
An e-commerce company exports its daily sales report as a CSV file.

```python
students.to_csv("students.csv", index=False)
print("CSV file created successfully.")
```

```python
df = pd.read_csv("students.csv")

print(df)
```

### Excel Files
Excel files store data in worksheets with the extensions .xlsx or .xls.
Example:

A company's HR department stores employee records in Excel files.

```python
students.to_excel("students.xlsx", index=False)
print("Excel file created successfully.")
```

### AI/ML Usage
Business datasets are often received in Excel format before being cleaned and converted for machine learning.

### JSON Files
JSON (JavaScript Object Notation) stores data as key-value pairs.

It is widely used in APIs and web applications.

Example:

A weather API returns temperature, humidity, and wind speed in JSON format.

```python
students.to_json(
    "students.json",
    orient="records",
    indent=4
)

print("JSON file created.")
```

```python
df = pd.read_json("students.json")
print(df)
```

Parquet is a column-based storage format that is highly compressed and optimized for big data processing.
Example:

Netflix stores large user-viewing datasets in Parquet because it loads faster and uses less storage.

### Pickle Files
Pickle stores Python objects in a binary format.
Unlike CSV or Excel, Pickle preserves the original Python object.
Example:
A data scientist saves a cleaned DataFrame to avoid repeating preprocessing every time.

```python
students.to_pickle("students.pkl")
print("Pickle file created.")
```

```python
df = pd.read_pickle("students.pkl")
print(df)
```

### SQL Tables
SQL databases store data in tables consisting of rows and columns.
Pandas can both read from and write to SQL databases.
Example:

Banks store customer transactions in SQL databases.

```python
import sqlite3

#Create Database Connection
connection = sqlite3.connect("students.db")
```

```python
students.to_sql(
    "student_table",
    connection,
    if_exists="replace",
    index=False
)

print("Table created successfully.")
```

```python
df = pd.read_sql(
    "SELECT * FROM student_table",
    connection
)

print(df)
```

```python
connection.close()
```

### Reading Multiple Files
Sometimes data is split across multiple files, and we need to combine them into one DataFrame.

Example:

A supermarket stores one sales CSV file for each month.

January.csv

February.csv

March.csv

The analyst combines them into one dataset.

```python
sales1 = pd.DataFrame({
    "Product": ["Laptop", "Phone"],
    "Sales": [100, 200]
})

sales2 = pd.DataFrame({
    "Product": ["Tablet", "Watch"],
    "Sales": [150, 120]
})

sales1.to_csv("sales1.csv", index=False)
sales2.to_csv("sales2.csv", index=False)

print(sales1)
print()
print(sales2)
```

```python
#Read Multiple Files
df1 = pd.read_csv("sales1.csv")
df2 = pd.read_csv(".csv")
combined = pd.concat([df1, df2], ignore_index=True)
print(combined)
```
