### NumPy Random Module
#### Random Numbers
Definition: Random numbers are values generated in a way that makes them unpredictable or appear randomly selected.
NumPy's np.random.random() generates random floating-point numbers between 0.0 and 1.0.

```python
import numpy as np

print(np.random.random())
print(np.random.random(5))
print(np.random.random((2, 3)))

#np.random.random() — generates one random number.
#np.random.random(5) — generates five random numbers.
#np.random.random((2, 3)) — generates a 2D array with 2 rows and 3 columns.
```

### Random Seed
Definition: A random seed is a starting value used by a pseudorandom number generator. It allows you to reproduce the same sequence of random numbers.

```python
import numpy as np

np.random.seed(42)
print(np.random.random(5))

np.random.seed(42)
print(np.random.random(5))
```

#### AI/ML Usage
Random seeds make machine learning experiments reproducible.

### Random Integers
Generates random integer values within a specified range.

Example
A teacher wants to generate 10 random roll numbers for selecting students to answer questions.

```python
roll_numbers = np.random.randint(1,51,size=10)
print(roll_numbers)
```

### Random Choice
Selects random values from a given list or array.

It can also generate random values according to specified probabilities.

Example
An e-commerce company wants to randomly choose one customer for a free gift.

```python
customers = ["Thamarai","Ravi","Priya","John","Sam"]

winner = np.random.choice(customers)
print(np.random.choice(customers, size=2))
print(winner)
```

AI/ML Usage:
Used in bootstrap sampling, mini-batch selection, and recommendation systems.

### Shuffle
Shuffle randomly rearranges the elements of an array.

It changes the original array.

Example
Suppose exam questions are arranged in order. Before giving the paper to students, the questions are shuffled.

```python
import numpy as np
numbers = np.array([10, 20, 30, 40, 50])
np.random.shuffle(numbers)
print(numbers)
```

### Permutation
permutation() returns a randomly permuted sequence or array without modifying the original array.

```python
import numpy as np
numbers = np.array([10, 20, 30, 40, 50])
result = np.random.permutation(numbers)
print("Original:", numbers)
print("Permuted:", result)
```

### Random Distributions
A Random Distribution generates numbers that follow a specific statistical pattern instead of being completely random.

Some commonly used distributions are:

Uniform Distribution
Normal Distribution
Binomial Distribution
These are widely used in AI, ML, simulations, and statistical modeling.

### Uniform Distribution
Every value has an equal chance of being selected.

Example
Suppose an online shopping website wants to generate a random discount between 10% and 20%, where every value has an equal chance.

```python
discount = np.random.uniform(10,20,5)
print(discount)
```

### AI/ML Usage
Used in random simulations and weight initialization.

### B. Normal Distribution
Generates values that follow a bell-shaped curve.

Most values are close to the mean.

#### Example
The heights of students in a school usually follow a normal distribution.

```python
import numpy as np

heights = np.random.normal(
    loc=165,
    scale=7,
    size=10
)

print(heights)
#Most students are close to 165 cm, while very short and very tall students are less common.
```

### C. Binomial Distribution
Produces the number of successes in a fixed number of independent trials.

Example
A cricket player plays 10 balls. Each ball has a 60% chance of scoring a run. We want to know how many successful scoring shots occur.

```python
import numpy as np
runs = np.random.binomial(
    n=10,
    p=0.6,
    size=5
)
print(runs)
```

Each value represents the number of successful scoring shots out of 10 balls.

### AI/ML Usage
Used in binary classification, success/failure experiments, A/B testing, and probability modeling.
