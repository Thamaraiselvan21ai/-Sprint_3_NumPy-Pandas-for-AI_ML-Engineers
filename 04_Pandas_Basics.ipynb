1. What is Pandas?
Pandas is an open-source Python library used for working with structured or tabular data. It provides powerful data structures like Series and DataFrame to efficiently store, manipulate, analyze, and visualize data.

Pandas simplifies tasks such as:

Reading data from files
Cleaning missing values
Filtering data
Sorting data
Grouping data
Performing statistical analysis
Pandas is built on top of the NumPy library, making it fast and efficient.

Example
Imagine a school has a spreadsheet containing information about 1,000 students.

Instead of manually calculating averages or finding the highest mark, Pandas can perform these operations in just a few lines of code.

```python
import pandas as pd
data = {
    "Name": ["Arun", "Thamarai", "Kumar"],
    "Age": [25, 28, 30],
    "Salary": [25000, 35000, 40000]
}
df = pd.DataFrame(data)
print(df)
```

Why Pandas?
| Task | Pandas Functionality |
|---|---|
| Read data | Load CSV, Excel, and other supported formats |
| Clean data | Handle missing and duplicate values |
| Filter data | Select rows based on conditions |
| Analyze data | Calculate mean, median, sum, and statistics |
| Transform data | Create, modify, and delete columns |
| Combine data | Merge, join, and concatenate datasets |
| Save data | Export results to CSV or Excel |
| Prepare ML data | Clean and transform features before training |

### Series
A Series is a one-dimensional labeled array that can store values such as integers, strings, floats, and other Python objects.
Think of a Series as a single column of data with an index.

```python
import pandas as pd
ages = pd.Series([22, 25, 28, 30])
print(ages)
```

```python
marks = pd.Series(
    [85, 90, 78],
    index=["Arun", "Priya", "Kumar"]
)

print(marks)
print(marks["Priya"])
```

### DataFrame
A DataFrame is a two-dimensional labeled data structure consisting of rows and columns.
Think of a DataFrame as an Excel sheet or SQL table.

```python
import pandas as pd
data = {
    "Name": ["Arun", "Priya", "Kumar"],
    "Age": [22, 25, 28],
    "Marks": [85, 90, 78]
}
df = pd.DataFrame(data)
print(df)
```

```python
print(df.shape)
print(df.size)
print(df.ndim)
print(df.index)
print(df.columns)
print(df.dtypes)
```

### Creating DataFrames
There are several ways to create a DataFrame.
Method 1: From a dictionary

```python
import pandas as pd
data = {
    "Name": ["Arun", "Priya"],
    "Age": [22, 25]
}
df = pd.DataFrame(data)
print(df)
```

```python
# Method 2: From a list of lists
data = [
    ["Arun", 22],
    ["Priya", 25],
    ["Kumar", 28]
]

df = pd.DataFrame(data, columns=["Name", "Age"])
print(df)
```

```python
#Method 3: From a list of dictionaries
data = [
    {"Name": "Arun", "Age": 22},
    {"Name": "Priya", "Age": 25},
    {"Name": "Kumar", "Age": 28}
]

df = pd.DataFrame(data)
print(df)
```

### Reading Data
Reading data means importing data from external files into a Pandas DataFrame so that it can be analyzed and processed.

Pandas supports many file formats such as:

* CSV
* Excel
* JSON
* SQL Database
* Parquet

```python
import pandas as pd
students = pd.DataFrame({
    "ID": [101, 102, 103, 104],
    "Name": ["Aishu", "Rahul", "Priya", "John"],
    "Age": [22, 23, 21, 24],
    "Marks": [95, 88, 91, 84]
})

print(students)
```

```python
### Create a CSV File
students.to_csv("students.csv", index=False)
print("CSV file created successfully.")
```

```python
#Read CSV File
df = pd.read_csv("students.csv")
print(df)
```

```python
#Create an Excel File
students.to_excel("students.xlsx", index=False)
print("Excel file created successfully.")
```

```python
#Read Excel File
df = pd.read_excel("students.xlsx")
print(df)
```

```python
#Create a JSON File
students.to_json("students.json", orient="records", indent=4)
print("JSON file created successfully.")
```

```python
#Read JSON File
df = pd.read_json("students.json")
print(df)
```

8 Create a New DataFrame

```python
import pandas as pd
employees = pd.DataFrame({
    "Employee_ID": [1, 2, 3],
    "Name": ["John", "Sara", "David"],
    "Department": ["IT", "HR", "Finance"],
    "Salary": [50000, 45000, 60000]
})

print(employees)
```

```python
#Write to CSV
employees.to_csv("employees.csv", index=False)
print("employees.csv saved successfully.")
```

```python
#Write to Excel
employees.to_excel("employees.xlsx", index=False)
print("employees.xlsx saved successfully.")
```

```python
#Write to JSON
employees.to_json("employees.json", orient="records", indent=4)
print("employees.json saved successfully.")
```

```python
#Verify the Saved Files
csv_data = pd.read_csv("employees.csv")
excel_data = pd.read_excel("employees.xlsx")
json_data = pd.read_json("employees.json")
print(csv_data)

print(excel_data)

print(json_data)
```

```python
import pandas as pd

data = {
    "Name": ["Arun", "Priya", "Kumar", "Divya"],
    "Age": [22, 25, 28, 24],
    "Salary": [25000, 35000, 40000, 30000],
    "Department": ["IT", "HR", "IT", "Finance"]
}

df = pd.DataFrame(data)

# 1. Display the first two rows
print(df.head(2))

# 2. Display the shape
print(df.shape)

# 3. Display only the Name column
print(df["Name"])

# 4. Display Name and Salary
print(df[["Name", "Salary"]])

# 5. Display employees whose salary is greater than 30000
print(df[df["Salary"] > 30000])

# 6. Calculate the average salary
print(df["Salary"].mean())

# 7. Save the dataset
df.to_csv("employees.csv", index=False)
```
