```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50])
```

### 1. Sum — np.sum()
Definition: Adds all elements in an array.

```python
print(np.sum(arr))
```

```python
matrix = np.array([[1, 2, 3],
                   [4, 5, 6]])

print(np.sum(matrix))         # 21
print(np.sum(matrix, axis=0)) # [5 7 9] — column sums
print(np.sum(matrix, axis=1)) # [6 15] — row sums
```

AI/ML example: Calculate the total sales across all transactions or the total loss across a batch of predictions.

### 2. Mean — np.mean()
Definition: Calculates the arithmetic average of the elements.

```python
print(np.mean(arr))
```

### 3. Median — np.median()
Definition: Finds the middle value after sorting the data.

```python
print(np.median(arr))
```

```python
x = np.array([10, 20, 30, 40])
print(np.median(x))
```

### 4. Variance — np.var()
Definition: Measures how far the data values spread around their mean, using squared deviations.

```python
x = np.array([10, 20, 30, 40, 50])
print(np.var(x))
```

```python
print(np.var(x, ddof=1))
```

Interpretation: Higher variance means greater spread; lower variance means values are closer to the mean.

```python
print(np.std(arr))
```

### Minimum — np.min()
Definition: Returns the smallest value in an array.

```python
print(np.min(arr))
```

```python
#for matrix form
matrix = np.array([[10, 20],
                   [ 5, 40]])

print(np.min(matrix))         # 5
print(np.min(matrix, axis=0)) # [5 20]
print(np.min(matrix, axis=1)) # [10 5]
```

### 9. Argmax — np.argmax()
Definition: Returns the index of the largest element.

```python
x = np.array([40, 10, 30, 5, 20])
print(np.argmax(x))
```

```python
print(x[np.argmax(x)])
#AI/ML example: Find the predicted class with the highest probability.
```

```python
probabilities = np.array([0.10, 0.75, 0.15])
predicted_class = np.argmax(probabilities)
print(predicted_class)
#The model predicts class index 1 because its probability is 0.75.
```

### 10. Percentile — np.percentile()
Definition: Finds the value below which a specified percentage of observations falls, using NumPy's percentile calculation method.

```python
x = np.array([10, 20, 30, 40, 50])

print(np.percentile(x, 25)) 
print(np.percentile(x, 50))  
print(np.percentile(x, 75)) 
print(np.percentile(x, 90)) 
```

### Matrix multiplication — np.matmul() or @
Definition: Multiplies rows of the first matrix by columns of the second matrix and adds the products.

```python
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

print(np.matmul(A, B))
print(A @ B) #For matrix operations, @ is often the easiest syntax to read.
```

### Transpose — np.transpose() or .T
Definition: Swaps the rows and columns of a matrix.
Example:
\[
A=\begin{bmatrix}1&2&3\\4&5&6\end{bmatrix}
\]
After transposing:
\[
A^T=\begin{bmatrix}1&4\\2&5\\3&6\end{bmatrix}
\]

```python
import numpy as np
A = np.array([[1, 2, 3],
              [4, 5, 6]])

print(A.T)

print(np.transpose(A))
# Same output
```

### Basic linear algebra operations
NumPy provides numpy.linalg for common linear algebra calculations.
#### A. Element-wise addition and subtraction
These operations work on corresponding elements.

```python
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

print(A + B)

print(A - B)
```

### Element-wise multiplication
Use * to multiply corresponding elements.

This is different from matrix multiplication using A @ B.

### C. Scalar multiplication
Multiplies every element by a single number.

```python
print(A * 2)
```

### Matrix determinant — np.linalg.det()
Definition: Calculates a scalar value associated with a square matrix. A nonzero determinant indicates that the matrix is invertible.

##### For a 2*2 matrix:
det(A)=ad-bc

```python
A = np.array([[1, 2],
              [3, 4]])

print(np.linalg.det(A))
```

### Matrix inverse — np.linalg.inv()
Definition: Calculates the inverse of an invertible square matrix.

AA^{-1}=I
##### Here, \(I\) is the identity matrix.

```python
A = np.array([[1, 2],
              [3, 4]])

print(np.linalg.inv(A))
```

### F. Identity matrix — np.eye()
Creates a matrix with ones on the main diagonal and zeros elsewhere.

```python
print(np.eye(3))
```

### Solve linear equations — np.linalg.solve()
Suppose we have:
\[
2x+y=5
\]

\[
x+3y=6
\]

Represent the equations as \(Ax=b\) and solve for \(x\) and \(y\).

```python
A = np.array([[2, 1],
              [1, 3]])

b = np.array([5, 6])

solution = np.linalg.solve(A, b)

print(solution)
```

### H. Vector magnitude — np.linalg.norm()
Definition: Measures the length or magnitude of a vector.
For a vector \([3,4]\):
\[
\|\mathbf{x}\|_2=\sqrt{3^2+4^2}=5
\]
